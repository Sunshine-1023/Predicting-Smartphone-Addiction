# 03 Final report notebook

Only reads **public** summaries under `reports/`.
After runs are selected, export charts and keep `final_report.md` in sync.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

from smartphone_addiction.paths import project_root

ROOT = project_root()
REP = ROOT / "reports"
FIG = REP / "figures"
FIG.mkdir(parents=True, exist_ok=True)

summary_path = REP / "experiment_summary.csv"
subs_path = REP / "submissions.csv"
summary = pd.read_csv(summary_path) if summary_path.stat().st_size else pd.DataFrame()
subs = pd.read_csv(subs_path) if subs_path.stat().st_size else pd.DataFrame()
display(summary)
display(subs)
```

```python
auc_col = "oof_auc" if "oof_auc" in summary.columns else "oof_auc_mean"
if not summary.empty and auc_col in summary.columns:
    plot_df = summary.dropna(subset=[auc_col]).copy()
    if not plot_df.empty:
        ax = plot_df.plot(
            x="run_id", y=auc_col, kind="bar", figsize=(8, 4), legend=False
        )
        ax.set_ylabel("OOF AUC")
        ax.set_title("Selected experiment OOF AUC")
        plt.tight_layout()
        plt.savefig(FIG / "final_oof_auc.png", dpi=120)
        plt.close()
    else:
        print(f"No {auc_col} values yet.")
else:
    print("experiment_summary.csv has no completed rows yet.")
```

```python
# Optional importance preview if a CSV was exported under a run
# Edit RUN_ID after training + importance export.
RUN_ID = ""
if RUN_ID:
    imp = Path("artifacts/runs") / RUN_ID / "importance"
    files = sorted(imp.glob("permutation_*.csv")) if imp.exists() else []
    print(files)
    if files:
        frame = pd.read_csv(files[0]).head(20)
        display(frame)
        ax = frame.plot(x="feature", y="importance_mean", kind="barh", figsize=(8, 6), legend=False)
        ax.invert_yaxis()
        plt.tight_layout()
        plt.savefig(FIG / f"final_importance_{RUN_ID}.png", dpi=120)
        plt.close()
else:
    print("Set RUN_ID to preview permutation importance.")
```
