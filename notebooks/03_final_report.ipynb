```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..') if not Path('reports').exists() and Path('../reports').exists() else Path('.')
REP = ROOT / 'reports'
FIG = REP / 'figures'
FIG.mkdir(parents=True, exist_ok=True)

summary = pd.read_csv(REP / 'experiment_summary.csv') if (REP / 'experiment_summary.csv').stat().st_size else pd.DataFrame()
subs = pd.read_csv(REP / 'submissions.csv') if (REP / 'submissions.csv').stat().st_size else pd.DataFrame()
display(summary)
display(subs)
```

```python
if not summary.empty and 'oof_auc_mean' in summary.columns:
    ax = summary.plot(x='run_id', y='oof_auc_mean', kind='bar', figsize=(8, 4), legend=False)
    ax.set_ylabel('OOF AUC')
    ax.set_title('Selected experiment OOF AUC')
    plt.tight_layout()
    plt.savefig(FIG / 'final_oof_auc.png', dpi=120)
    plt.close()
```
